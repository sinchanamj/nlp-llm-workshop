#linear regression

Linear Regression is a Machine Learning algorithm used to predict a numerical value.

```python
import numpy as np
import matplotlib.pyplot as plt
import sklearn
from sklearn.linear_model import LinearRegression
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([50,55,65,70,85])
model=LinearRegression()
model.fit(x,y)
xs=np.linspace(0.5,6.5,100).reshape(-1,1)
plt.scatter(x,y,color="red",s=70, zorder=3, label="training data")
plt.plot(xs,model.predict(xs),color="steelblue",lw=2,label="fitted line")
plt.scatter([6],model.predict([[6]]),color="teal",s=90,zorder=3,label="predictiom")
for xi,yi in zip(x.flatten(),y):
  plt.plot([xi,xi],[yi,model.predict([[xi]])[0]],"k--",lw=0.8)
plt.xlabel("hours studied"); plt.ylabel("exam score")
plt.legend();plt.title("linear regression");plt.show()
```

```python
from torch import nn
import torch
import matplotlib
from matplotlib import pyplot as plt
```

```python
x=torch.arange(0,1,0.02).unsqueeze(dim=1)
print(x)
```

```python
y=(x*0.7)+0.3
```

```python
len(x)
```

```python
len(y)
```

```python
split=int(0.8*len(x))
x_train,y_train=x[:split],y[:split]
x_test,y_test=x[split:],y[split:]
```

```python
split
```

```python
len(x_train),len(y_train),len(x_test),len(y_test)
```

```python
plt.scatter(x_train,y_train,c="b",s=4,label="training data")
plt.scatter(x_test,y_test,c="g",s=4,label="testing data")
plt.legend()
plt.show()
```

```python
class LinearReg(nn.Module):
  def __init__(self):
    super().__init__()

    self.weights=nn.Parameter(
        torch.rand(1,dtype=torch.float),requires_grad=True
    )
    self.bias=nn.Parameter(
        torch.rand(1,dtype=torch.float),requires_grad=True
    )
  def forward(self,x:torch.Tensor):
    return (x*self.weights)+self.bias
```

```python
torch.manual_seed(42)
lr=LinearReg()
```

```python
list(lr.parameters())
```

```python
lr.state_dict()
```

```python
with torch.inference_mode():
  y_preds=lr(x_test)
```

```python
plt.scatter(x_train,y_train,c="b",s=5,label="training data")
plt.scatter(x_test,y_test,c="g",s=5,label="testing data")
plt.scatter(x_test,y_preds,c="r",s=5,label="prediction")
plt.legend()
plt.show()
```

```python
loss=torch.nn.L1Loss()
```

```python
optim=torch.optim.Adam(params=lr.parameters(),lr=0.01)
```

```python
torch.manual_seed(42)
epochs=109
for epoch in range(epochs):
  lr.train()
  y_preds=lr(x_train)
  loss_fn=loss(y_preds,y_train)
  optim.zero_grad()
  loss_fn.backward()
  optim.step()
  lr.eval()
  with torch.inference_mode():
    test_preds=lr(x_test)
    test_loss=loss(test_preds,y_test)
  print(f"EPOCH: {epoch} | LOSS: {loss_fn} | TEST LOSS: {test_loss}")
```

#SAMPLE RAW TEXT

```python
raw_text="""Hello!! I'm So Happy. This Product Is AMAZING😘😘😘!!
I've been using it for 2 weeks and it's been running smoothly.
visit https:example.com for more info.
<p>best purchase ever</p>"""
```

```python
print("=" *60)
print("raw_text")
print("=" * 60)
print(raw_text)
```

step 1: lowercasing

```python
text_lower=raw_text.lower()
print("=" * 60)
print("text_lower")
print("=" * 60)
print(text_lower)
```

step 2: remove html tags

```python
text_no_html=re.sub(r'<.*?>','',text_lower)
print("=" * 60)
print("text_no_html")
print("=" * 60)
print(text_no_html)
```

step 3: remove urls

```python
text_no_urls=re.sub(r'http\S+','',text_no_html)
print("=" * 60)
print("text_no_urls")
print("=" * 60)
print(text_no_urls)
```

step 4: remove emojis

```python
text_no_emojis=text_no_urls.encode('ascii','ignore').decode('ascii')
print("=" * 60)
print("text_no_emojis")
print("=" * 60)
print(text_no_emojis)
```

step 5: numbers

```python
text_no_num=re.sub(r'\d+','',text_no_emojis)
print("=" * 60)
print("text_no_num")
print("=" * 60)
print(text_no_num)
```

step 6:punctuation

```python
text_no_punc=text_no_num.translate(str.maketrans('','',string.punctuation))
print("=" * 60)
print("text_no_punc")
print("=" * 60)
print(text_no_punc)

```

step 7: remove white space

```python
text_no_white_space=re.sub(r'\s+',' ',text_no_punc)
print("=" * 60)
print("text_no_white_space")
print("=" * 60)
print(text_no_white_space)
```
